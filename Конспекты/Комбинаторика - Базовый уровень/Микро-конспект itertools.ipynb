Вы уже скорее всего ознакомились с комбинаторикой из прошлого конспекта, и у вас (если вы ещё не знакомы с библиотекой itertools) возникнет вопрос: "А как ей оперировать в программировании, а конкретно в питоне?" Ведь и вправду, оперировать все эти функции через циклы было бы очень неудобно и муторно. Но как раз для такой цели существует библиотека itertools, о которой мы коротко здесь поведаем.

Функция Product возвращает сочетания указанных элементов


```python
#Перебор всех вариантов
import itertools

a = [1,2,3] #элементы, которые мы сочетаем
#Вывод результатов данной функции, и оперирование с ними происходит спомощью цикла, ввиду особенностей структуры возвращаемых данных
#Выглядит это так:
for elem in itertools.product(a,repeat=2):
  print(elem)
#параметр repeat отвечает за максимальную(!!!) длину сочетания, тоесть сколько максимально(!!!) элементов будет сочетаться
#так же элементы могут сочетаться с самими собой
#функция учитывает порядок записи чисел, по этому два одинаковых сочетания могут считаться как два разных
```

Функция Permutations возвращает все возможные перестановки в указанном ряду элементов

```python
#Функция берет в качестве своих значений аргументы iterables - последовательность в которую тасуются элементы и r - длина возвращаемой последовательности
for elem in itertools.permutations("abc",3):
  print(elem)
```

```python
#Что будет если поставить r больше длины последовательности?
for elem in itertools.permutations("abc",4):
  print(elem)
```

```python
#А если меньше?
for elem in itertools.permutations("abc",2):
  print(elem)
```

Функция Combinations возвращает комбинации указанных элементов. Отличается от сочетаний это тем, что комбинации не содержат повторений элементов

```python
#в функцию подаются два аргумента iterable и r. Аналогичны тому что было показано выше.
for elem in itertools.combinations([1,2,3],2):
  print(elem)
```
