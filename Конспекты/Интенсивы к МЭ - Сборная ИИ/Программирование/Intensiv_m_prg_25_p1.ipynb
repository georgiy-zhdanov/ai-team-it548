# Рекурсии и сортировки

## Что такое рекурсия?

Рекурсия — это прием, когда функция вызывает саму себя для решения подзадачи меньшего размера. У любой рекурсии должно быть два элемента:

1.  **Базовый случай (Base case):** Условие остановки, когда ответ понятен без вычислений.
2.  **Шаг рекурсии:** Вызов функции с новыми аргументами.


## Пример 1: Числа Фибоначчи

Классический пример. Числа Фибоначчи определяются так:
$$F(0) = 0, \quad F(1) = 1, \quad F(n) = F(n-1) + F(n-2)$$

### Наивная реализация

Если мы запишем формулу «в лоб», алгоритм будет работать экспоненциально долго — $O(2^n)$. Это происходит потому, что одни и те же значения вычисляются много раз.

### Мемоизация (Memoization)

Чтобы ускорить процесс, мы можем запоминать уже вычисленные значения в словаре (хеш-таблице) или массиве. Это сокращает сложность до $O(n)$.


````python
import time

# 1. Наивная рекурсия (Медленно!)
def fib_naive(n):
    if n <= 1:
        return n
    return fib_naive(n-1) + fib_naive(n-2)

# 2. Рекурсия с мемоизацией (Быстро!)
# Используем словарь для хранения вычисленных значений
memo = {}

def fib_memo(n):
    if n in memo:
        return memo[n]

    if n <= 1:
        return n

    # Запоминаем результат перед возвратом
    memo[n] = fib_memo(n-1) + fib_memo(n-2)
    return memo[n]

# Тестирование
start = time.time()
print(f"Fib(30) naive: {fib_naive(30)}")
print(f"Время naive: {time.time() - start:.4f} сек")

start = time.time()
memo = {} # Очистим кэш
print(f"Fib(30) memo:  {fib_memo(30)}")
print(f"Время memo:  {time.time() - start:.4f} сек")
````

## Пример 3: Перебор вариантов (Backtracking)

Рекурсия идеально подходит для перебора всех вариантов (брутфорс). Например, генерация всех двоичных строк длины $N$.

Принцип:

1.  Ставим на текущую позицию `0`, вызываем рекурсию для следующей позиции.
2.  Ставим на текущую позицию `1`, вызываем рекурсию.
3.  Когда длина строки равна $N$ — выводим и завершаем ветку.

**Оптимизация (Отсечение):**
Если нам нужны строки с **ровно K единицами**, мы можем не заходить в ветки рекурсии, где единиц уже больше $K$ или где их слишком мало, чтобы набрать $K$ оставшимися битами.


````python
def generate_binary_strings(n, k, current_string="", ones_count=0):
    """
    n: целевая длина
    k: требуемое количество единиц
    current_string: текущая собранная строка
    ones_count: сколько единиц уже поставили
    """
    # Базовый случай: строка собрана
    if len(current_string) == n:
        if ones_count == k:
            print(current_string)
        return

    # ОТСЕЧЕНИЯ (Pruning) - ускоряют перебор
    # 1. Если уже поставили больше k единиц - нет смысла продолжать
    if ones_count > k:
        return
    # 2. Если оставшихся мест не хватит, чтобы добрать до k единиц
    # Осталось мест: n - len(current_string)
    # Нужно еще единиц: k - ones_count
    if (n - len(current_string)) < (k - ones_count):
        return

    # Шаг 1: Попробовать поставить '0'
    generate_binary_strings(n, k, current_string + "0", ones_count)

    # Шаг 2: Попробовать поставить '1'
    generate_binary_strings(n, k, current_string + "1", ones_count + 1)

print("Двоичные строки длины 4 с 2 единицами:")
generate_binary_strings(4, 2)
````

## Сортировка слиянием (Merge Sort)

Это алгоритм "Разделяй и властвуй".

1.  **Разделяем** массив на две половины.
2.  **Рекурсивно** сортируем каждую половину.
3.  **Сливаем** (Merge) две отсортированные половины в один массив.

Сложность: $O(N \log N)$ всегда. Это стабильная сортировка (не меняет порядок равных элементов).

## Сортировка подсчетом (Counting Sort)

Если мы знаем, что числа в массиве лежат в небольшом диапазоне (например, от 0 до 100), мы можем не сравнивать их друг с другом.
Мы просто считаем, сколько раз встречается каждое число, а потом выводим их нужное количество раз.

Сложность: $O(N + M)$, где $M$ — размер диапазона значений.

````python
def merge_sort(arr):
    # Базовый случай: массив длины 0 или 1 уже отсортирован
    if len(arr) <= 1:
        return arr

    # 1. Разделение
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])   # Рекурсивный вызов
    right_half = merge_sort(arr[mid:])  # Рекурсивный вызов

    # 2. Слияние
    return merge(left_half, right_half)

def merge(left, right):
    sorted_arr = []
    i = j = 0

    # Пока в обоих массивах есть элементы, выбираем меньший
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            sorted_arr.append(left[i])
            i += 1
        else:
            sorted_arr.append(right[j])
            j += 1

    # Дописываем остатки (если есть)
    sorted_arr.extend(left[i:])
    sorted_arr.extend(right[j:])

    return sorted_arr

# Тест
arr = [38, 27, 43, 3, 9, 82, 10]
print(f"Исходный: {arr}")
print(f"Отсортированный: {merge_sort(arr)}")
````
